```python
import pandas as pd
import numpy as np
```

```python
import os
movies_path = os.path.join(os.path.dirname(os.getcwd()), "DATA", "movies.csv")
df = pd.read_csv(movies_path)
df.head(10)
```

```python
from IPython.display import display, Markdown

def explorer_dataset(df):
    # 1. Dimensions et Types (Résumé rapide)
    display(Markdown("### 📊 Aperçu Structurel"))
    print(f"Dimensions : {df.shape[0]} lignes et {df.shape[1]} colonnes")

    # 2. Les 5 dernières lignes (Format Tableau)
    display(Markdown("---"))
    display(Markdown("### 🔍 Dernières lignes"))
    display(df.tail())

    # 3. Échantillon aléatoire (Idéal pour détecter les erreurs de saisie)
    display(Markdown("---"))
    display(Markdown("### 🎲 Échantillon aléatoire (5 lignes)"))
    display(df.sample(5))

    # 4. Santé des données (Types et Valeurs manquantes)
    display(Markdown("---"))
    display(Markdown("### 🛠 Santé et Types des colonnes"))
    # On crée un petit DataFrame pour résumer l'info proprement
    info_df = pd.DataFrame({
        'Type': df.dtypes,
        'Valeurs Manquantes': df.isna().sum(),
        '% Manquant': (df.isna().sum() / len(df) * 100).round(2).astype(str) + '%'
    })
    display(info_df)

    # 5. Statistiques descriptives
    display(Markdown("---"))
    display(Markdown("### 📈 Statistiques des variables numériques"))
    display(df.describe())



# Appel de la fonction
explorer_dataset(df)

print(f"Films avec budget à 0 : {(df['budget'] == 0).sum()}")
print(f"Films avec revenu à 0 : {(df['revenue'] == 0).sum()}")

# Pour voir le pourcentage d'anomalies
print(f"Pourcentage de budget 'manquant' : { (df['budget'] == 0).mean() * 100:.2f}%")
```
